```python
import pandas as pd
import numpy as np
import joblib 
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
```

```python
def load_and_explore_dataset(file_path):
    print("=" * 60)
    print("LOAD AND EXPLORE DATA")
    print("=" * 60)

    pd.set_option('display.float_format', lambda x: f'{x:.2f}')

    df = pd.read_csv(file_path)
    print("Shape of Dataset:")
    print(df.shape)
    print("\nChecking for Missing Value:")
    print(df.isnull().sum())
    print("\nFirst 5 Row:")
    print(df.head())
    print("\nDescriptive Statistics:")
    print(df.describe())
    print("\nDataset Info:")
    print(df.info())
    print("\nExited Distribution")
    print(df["Exited"].value_counts)
    print("\nExited Percentage Distribution")
    print(df["Exited"].value_counts(normalize=True) * 100)

    return df

    
```

```python
#Seperating categorical data from numerical ones

def identify_feature(df):
    print("\n" + "=" * 60)
    print("FEATURES IDENTIFICATION")
    print("=" * 60)

    #features to drop
    feature_to_drop = [
        "CustomerId",
        "Surname"
    ]

    #numerical columns
    numerical_columns = [
        "CreditScore",
        "Age",
        "Tenure",
        "EstimatedSalary",
        "Balance",
        "NumOfProducts",
        "ServiceRating"
    ]

    
    #categories columns
    categorical_columns = [
        "Geography",
        "Gender",
        "MasterCard"
        "IsActiveMember"
    ]

    print("\nFeature to Drop", feature_to_drop)
    print("\nNumerical Columns", numerical_columns)
    print("\nCategorical Columns", categorical_columns)

    return feature_to_drop, numerical_columns, categorical_columns
    

    
```

```python
def prepare_data(df, feature_to_drop):
    print("\n" + "=" * 60)
    print("PREPARED FEATURE DATASET")
    print("=" * 60)

    #drop unnecessary columns
    df_model = df.drop(columns=feature_to_drop)

    #Specify Label
    #Feature Label
    X = df_model.drop('Exited', axis=1)

    #Target Label
    y = df_model['Exited']

    print("\nFeature Shape:", X.shape)
    print("\nTarget Shape:", y.shape)
    print("\nFeature Columns:", list(X.columns))

    return X, y, list(X.columns)
    
```

```python
#OneHotEncoder: Converts categorical data into numerical columns of 1s and 0s. 
    #drop='first': removes dummy used in transformation to avoid perfect correlation
    #sparse_output: returns readdable Numpy Array
    #handle_unknown: 
#Pipeline: combines multiple sequencial values blocks into one object. It helps avoid data leakages
#ColumnTransformer: Apply different preprocessing steps to different columns at the same time
#StandardScale:

def create_processing_pipeline(numerical_columns, categorical_columns):
    print("\n" + "=" * 60)
    print("CREATE PREPROCESSING PIPELINE")
    print("=" * 60)

    # Numerical feature pipeline
    numerical_pipeline = Pipeline([
        ('Scaler', StandardScaler())
    ])

    # Categorical feature pipeline
    categories_pipeline = Pipeline([
        ('onehot', OneHotEncoder(drop="first", sparse_output=False, handle_unknown='ignore'))
    ])

    #Column Preprocessor
    preprocessor = ColumnTransformer([
        ('num', numerical_pipeline, numerical_columns),
        ('cat', categories_pipeline, categorical_columns)    
    ])

    return preprocessor
```

```python
def main():
    file_path = "data/cleaned_bank_customer_churn.csv"

    df = load_and_explore_dataset(file_path)

    feature_to_drop, numerical_columns, categorical_columns = identify_feature(df)

    X, y, feature_columns = prepare_data(df, feature_to_drop)

    preprocessor = create_processing_pipeline(numerical_columns, categorical_columns)
```

```python
if __name__ == "__main__":
    main()
```
